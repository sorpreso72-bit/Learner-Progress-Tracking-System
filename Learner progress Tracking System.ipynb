```python
import json
import os

FILE_NAME = "learners.json"


# Load saved learners
def load_learners():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return {}


# Save learners
def save_learners(learners):
    with open(FILE_NAME, "w") as file:
        json.dump(learners, file, indent=4)


# Calculate grade
def calculate_grade(average):
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    elif average >= 40:
        return "E"
    else:
        return "F"


# Add a learner
def add_learner(learners):
    print("\n--- Add Learner ---")

    learner_id = input("Enter learner ID: ")

    if learner_id in learners:
        print("A learner with this ID already exists.")
        return

    name = input("Enter learner name: ")
    surname = input("Enter learner surname: ")
    grade = input("Enter learner's class/grade: ")

    learners[learner_id] = {
        "name": name,
        "surname": surname,
        "class": grade,
        "subjects": {}
    }

    save_learners(learners)

    print("Learner added successfully!")


# Add subject and mark
def add_mark(learners):
    print("\n--- Add Subject Mark ---")

    learner_id = input("Enter learner ID: ")

    if learner_id not in learners:
        print("Learner not found.")
        return

    subject = input("Enter subject: ")

    try:
        mark = float(input("Enter mark (0-100): "))

        if mark < 0 or mark > 100:
            print("Mark must be between 0 and 100.")
            return

    except ValueError:
        print("Please enter a valid number.")
        return

    learners[learner_id]["subjects"][subject] = mark

    save_learners(learners)

    print("Mark added successfully!")


# Calculate average
def calculate_average(subjects):
    if not subjects:
        return 0

    return sum(subjects.values()) / len(subjects)


# View learner progress
def view_learner(learners):
    print("\n--- Learner Progress ---")

    learner_id = input("Enter learner ID: ")

    if learner_id not in learners:
        print("Learner not found.")
        return

    learner = learners[learner_id]

    print("\nLearner Information")
    print("-------------------")
    print(f"ID: {learner_id}")
    print(f"Name: {learner['name']} {learner['surname']}")
    print(f"Class: {learner['class']}")

    print("\nSubjects and Marks")
    print("------------------")

    if not learner["subjects"]:
        print("No marks recorded.")
        return

    for subject, mark in learner["subjects"].items():
        print(f"{subject}: {mark}%")

    average = calculate_average(learner["subjects"])
    grade = calculate_grade(average)

    print("\nOverall Performance")
    print("-------------------")
    print(f"Average: {average:.2f}%")
    print(f"Grade: {grade}")


# View all learners
def view_all_learners(learners):
    print("\n--- All Learners ---")

    if not learners:
        print("No learners registered.")
        return

    for learner_id, learner in learners.items():

        average = calculate_average(learner["subjects"])
        grade = calculate_grade(average)

        print(
            f"ID: {learner_id} | "
            f"Name: {learner['name']} {learner['surname']} | "
            f"Class: {learner['class']} | "
            f"Average: {average:.2f}% | "
            f"Grade: {grade}"
        )


# Search for a learner
def search_learner(learners):
    print("\n--- Search Learner ---")

    search = input("Enter learner ID or surname: ").lower()

    found = False

    for learner_id, learner in learners.items():

        if (
            search == learner_id.lower()
            or search in learner["surname"].lower()
        ):
            average = calculate_average(learner["subjects"])
            grade = calculate_grade(average)

            print("\nLearner Found")
            print("----------------")
            print(f"ID: {learner_id}")
            print(f"Name: {learner['name']} {learner['surname']}")
            print(f"Class: {learner['class']}")
            print(f"Average: {average:.2f}%")
            print(f"Grade: {grade}")

            found = True

    if not found:
        print("No learner found.")


# Main program
def main():
    learners = load_learners()

    while True:

        print("\n==============================")
        print(" LEARNER PROGRESS TRACKER")
        print("==============================")
        print("1. Add Learner")
        print("2. Add Subject Mark")
        print("3. View Learner Progress")
        print("4. View All Learners")
        print("5. Search Learner")
        print("6. Exit")
        print("==============================")

        choice = input("Choose an option: ")

        if choice == "1":
            add_learner(learners)

        elif choice == "2":
            add_mark(learners)

        elif choice == "3":
            view_learner(learners)

        elif choice == "4":
            view_all_learners(learners)

        elif choice == "5":
            search_learner(learners)

        elif choice == "6":
            print("\nThank you for using the Learner Progress Tracking System!")
            break

        else:
            print("Invalid option. Please choose 1-6.")


if __name__ == "__main__":
    main()
```
